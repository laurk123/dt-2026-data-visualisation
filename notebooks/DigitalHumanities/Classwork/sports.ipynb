```python
# can use to work with tabular data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() #Show the first few rows
df.describe() #Show some basic statistics
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visiualizations
df.plot(kind="bar", x="Name",y="Years_of_practice", legend=False)
plt.title("Years of practice per Athlete")
plt.ylabel("Years of practice")
plt.show
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visiualizations
df.plot(kind="bar", x="Name",y="Age")
plt.title("Age per Athlete")
plt.ylabel("Age of practice")
plt.show
```

```python

df.plot(kind="pie",y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Age of athletes")
plt.show()
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60, alpha=0.4)
#s=df["Years_of_practice"]*60) This means if student has more years of practice their bubble is bigger
# "alpha=0.6 makes the bubble more transparent 
plt.title("Age vs Years of practice")
plt.xlabel("Age")
plt.ylabel("Years of practice")
```

```python
import seaborn as sns #use this to create more clean 
sns.boxplot(data=df, y="Age")
plt.title("Distribution of athletes age")
plt.show
```

```python
plt.Figure(figsize=(8,5))
df[["Age","Years_of_practice"]].boxplot()
plt.title("Box plot for age and years of practice")
plt.show
```

```python
avg_age= df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
plt.show
```

```python
avg_age= df.groupby("Sport")["Years_of_practice"].mean()
avg_age.plot(kind="bar")
plt.show
```
