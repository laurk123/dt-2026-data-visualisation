Create a chart showing who puts in the most effort. Effort can be found from number of hours. (For example Student A has the highest study hours meaning they put in the most effort.)

```python
import pandas as pd
df = pd.read_csv("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt

avg_hours_studied = df.groupby("Student")["Hours_Studied"].mean()
avg_hours_studied.sort_values(ascending=False).plot(kind="bar")
plt.title("Average hours studied per student")
plt.ylabel("Average hours studied")
plt.xlabel("Student")
plt.tight_layout()
plt.show()
```

```python
plt.scatter(df["Grade"], df["Hours_Studied"], s=df["Hours_Studied"], alpha=0.4)
```

```python
import seaborn as sns
plt.Figure(figsize=(8,5))
df[["Grade","Hours_Studied"]].boxplot()
plt.title("Distribution of study grade and hours studied")
plt.show
```
