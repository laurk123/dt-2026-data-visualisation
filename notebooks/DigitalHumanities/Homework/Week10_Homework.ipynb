```python
text = "Cat sits by the window. She watches birds from the window."
words = text.split()

print("First three words:", words[:3])
print("Number of words:", len(words))
```

i used split() to break my sentences into words and i printed the first three words and used len() to count all of the words

```python
import urllib.request

url = "https://raw.githubusercontent.com/laurk123/homework-week-10/refs/heads/main/essay.txt"
text = urllib.request.urlopen(url).read().decode("utf-8")

words = text.split()
print("First three words:", words[:3])
print("Number of words:", len(words))
```

i loaded my Hong kong essay from github. i once again used split() to separate the words and len() to count them.

```python
from wordcloud import WordCloud

cloud = WordCloud(background_color="white").generate(text)
cloud.to_image()
```

i used python to make a world cloud from my essayy. "Hong" and "Kong" stands out the most same with "financial", "international" and "market". These words fit the main topic of my essay because i wrote about Hong Kongs role in international fiance.
